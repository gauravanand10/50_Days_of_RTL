```python
import numpy as np
from google.colab import files

def generate_sine_coe(file_name, amplitude, num_samples, bit_width):

    # Generate sine wave
    theta = np.linspace(0, 2 * np.pi, num_samples, endpoint=False)
    sine_wave = amplitude * np.sin(theta)

    # Convert to unsigned fixed-point values
    max_val = (2 ** bit_width) - 1
    offset = max_val // 2

    unsigned_values = np.round(sine_wave + offset).astype(int)
    unsigned_values = np.clip(unsigned_values, 0, max_val)

    # Print values
    print("=" * 60)
    print("Unsigned Sine Wave Samples")
    print("=" * 60)

    for i, value in enumerate(unsigned_values):
        print(f"Sample {i:3d} : {value}")

    # Create COE file
    coe_content = (
        "memory_initialization_radix=10;\n"
        "memory_initialization_vector=\n"
        + ",\n".join(map(str, unsigned_values))
        + ";\n"
    )

    # Save COE file
    with open(file_name, "w") as file:
        file.write(coe_content)

    # Print COE contents
    print("\n" + "=" * 60)
    print("Generated COE File")
    print("=" * 60)
    print(coe_content)

    print("\nCOE file generated successfully!")

    # Download the file automatically
    files.download(file_name)


# Parameters
output_file = "sine_wave_unsigned.coe"
amplitude = 127
num_samples = 256
bit_width = 8

# Generate COE File
generate_sine_coe(
    output_file,
    amplitude,
    num_samples,
    bit_width
)
```
